```python
IDS = [
  'registro-de-temperatura-365-dias',
  'pronostico-del-tiempo-a-5-dias'
]
```

```python
import requests

for id in IDS:
    API_URL = f"https://datos.gob.ar/api/3/action/package_show?id={id}"

    response = requests.get(API_URL).json()

    if response.get("success"):
        resources = response["result"]["resources"]
        print(f"Se encontraron {len(resources)} recursos en el dataset:\n")
        
        for res in resources:
            print(f"📌 Nombre: {res.get('name')}")
            print(f"   Formato: {res.get('format')}")
            print(f"   Resource ID: {res.get('id')}")
            print(f"   URL de descarga: {res.get('url')}\n")
```
